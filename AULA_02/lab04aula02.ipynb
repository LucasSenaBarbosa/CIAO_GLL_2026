```python
import random
import pandas as pd

# Definir os dados dos itens (nome, custo, utilidade)
items_data = {
    'item': ['Arroz', 'Feijão', 'Macarrão', 'Carne', 'Frango', 'Ovos', 'Leite', 'Pão', 'Queijo', 'Tomate', 'Cebola', 'Alface', 'Maçã', 'Banana', 'Café', 'Açúcar', 'Óleo', 'Sabonete', 'Pasta de Dente', 'Detergente'],
    'custo': [5.00, 7.00, 3.50, 25.00, 15.00, 10.00, 4.00, 6.00, 18.00, 3.00, 2.50, 4.00, 8.00, 7.00, 12.00, 4.50, 9.00, 3.00, 5.00, 4.00],
    'utilidade': [8, 9, 7, 10, 9, 8, 7, 6, 8, 5, 4, 3, 6, 6, 7, 5, 6, 4, 5, 3]
}

df_items = pd.DataFrame(items_data)

# Orçamento disponível
ORCAMENTO_MAXIMO = 100.00

print("Itens disponíveis:")
display(df_items)
print(f"Orçamento Máximo: R$ {ORCAMENTO_MAXIMO:.2f}")
```

```python
def gerar_solucao_aleatoria(num_itens):
    """Gera uma solução aleatória (vetor binário) para a lista de compras."""
    return [random.randint(0, 1) for _ in range(num_itens)]

def calcular_funcao_objetivo(solucao, df_items):
    """Calcula a utilidade total de uma solução."""
    utilidade_total = 0
    for i, incluir_item in enumerate(solucao):
        if incluir_item == 1:
            utilidade_total += df_items.loc[i, 'utilidade']
    return utilidade_total

def verificar_restricoes(solucao, df_items, orçamento_maximo):
    """Verifica se uma solução respeita as restrições de orçamento."""
    custo_total = 0
    itens_selecionados = []
    for i, incluir_item in enumerate(solucao):
        if incluir_item == 1:
            custo_total += df_items.loc[i, 'custo']
            itens_selecionados.append(df_items.loc[i, 'item'])

    violacao_orcamento = custo_total > orçamento_maximo

    return {
        'custo_total': custo_total,
        'violacao_orcamento': violacao_orcamento,
        'itens_selecionados': itens_selecionados
    }
```

```python
# Exemplo de uso:

# 1. Gerar uma solução aleatória
num_itens = len(df_items)
solucao_aleatoria = gerar_solucao_aleatoria(num_itens)

print(f"Solução Aleatória (vetor binário): {solucao_aleatoria}")

# 2. Calcular o valor da função objetivo dessa solução
utilidade = calcular_funcao_objetivo(solucao_aleatoria, df_items)
print(f"Utilidade Total da Solução: {utilidade}")

# 3. Verificar se a solução respeita as restrições
resultado_restricoes = verificar_restricoes(solucao_aleatoria, df_items, ORCAMENTO_MAXIMO)

custo_total = resultado_restricoes['custo_total']
violacao = resultado_restricoes['violacao_orcamento']
itens_selecionados = resultado_restricoes['itens_selecionados']

print(f"Custo Total da Solução: R$ {custo_total:.2f}")
print(f"Violação do Orçamento (Acima de R$ {ORCAMENTO_MAXIMO:.2f}): {violacao}")
print(f"Itens Selecionados: {', '.join(itens_selecionados) if itens_selecionados else 'Nenhum item selecionado'}")

if not violacao:
    print("\nEsta solução é FEASÍVEL (respeita o orçamento).")
else:
    print("\nEsta solução é INFEASÍVEL (excede o orçamento).")
```
